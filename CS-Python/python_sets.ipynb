```python
#To create a set using constructor
s=set(("apple","mango","banana"))
print(s)
```

```python
#To print items of a set 
for x in s:
    print(x)
```

```python
#To add items in set
s.add(1)
print(s)
```

```python
#To check if element exists in set
print("mango" in s)
```

```python
#To merge two sets
s1={1,2,3,4,7}#set
l=[1,2,3,4,5,6]#list
t=(4,5,6)#tuple
s.update(s1)
s.update(l)
s.update(t)
print(s)
```

```python
#To remove an item
s.remove(1)
print(s)
s.pop()
print(s)
s.clear()
print(s)
```

```python
#Union
s2={7,8,9}
v=s2.union(s1,l)
print(v)
```

```python
#Intersection
y=s2.intersection(s1)
print(y)
```

```python
#Difference
z=s1.difference(s2)
print(z)
```

```python
#To calculate length of sets
s={1,2,3}
print(len(s))
```
