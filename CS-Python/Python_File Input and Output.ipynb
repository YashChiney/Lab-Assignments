```python
#To read a document
f=open(R"C:\Users\Administrator\Desktop\data.txt","r")
df=f.read()
print(df)
```

```python
#To overwrite a document 
f=open(R"C:\Users\Administrator\Desktop\data.txt","w")
f.write("Hello , Good Morning")
f.close()
f=open(R"C:\Users\Administrator\Desktop\data.txt","r")
k=f.read()
print(k)
```

```python
#To continue writing in a document
f=open(R"C:\Users\Administrator\Desktop\data.txt","a")
f.write(" How are you!")
f.close()
f=open(R"C:\Users\Administrator\Desktop\data.txt","r")
d=f.read()
print(d)
```
